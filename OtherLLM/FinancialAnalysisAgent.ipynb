```python
import finnhub

# Setup client
finnhub_client = finnhub.Client(api_key=finnhub_io_api_key)
```

```python
# Get News 
from datetime import datetime
from datetime import datetime, timedelta

def get_sorted_news_headlines(ticker: str) -> str:
    """
    Fetches and processes news headlines for a given stock ticker within a 1-year date range.

    Args:
        ticker: The stock ticker symbol (e.g., 'TSLA').

    Returns:
        A multiline string of sorted news headlines.
    """
    # Compute date range
    today = datetime.now()
    one_year_ago = today - timedelta(days=365)
    _from = one_year_ago.strftime("%Y-%m-%d")
    to = today.strftime("%Y-%m-%d")

    # Fetch Company News Need to use _from instead of from to avoid conflict news
    news = finnhub_client.company_news(ticker, _from=_from, to=to)

    # Extract datetime and headline
    extracted_news = [(datetime.fromtimestamp(item['datetime']).date(), item['headline']) for item in news]

    # Sort by datetime
    sorted_news = sorted(extracted_news, key=lambda x: x[0])

    # Extract headlines from sorted_news
    sorted_news_headlines = [headline for _, headline in sorted_news]

    # Convert sorted_news_headlines to a multiline string
    sorted_news_headlines_multiline = "\n".join(sorted_news_headlines)

    return sorted_news_headlines_multiline

# Example usage
# print(get_sorted_news_headlines('TSLA'))
```

```python
# Define the prompt
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.schema import HumanMessage

def analyze_stock_sentiment(stock_symbol: str, news_article: str) -> str:
    """
    Analyzes the sentiment of a news article related to a specific stock.

    Args:
        stock_symbol: The ticker symbol of the stock (e.g., "CRCL").
        news_article: The text of the news article.

    Returns:
        A string indicating the sentiment (positive, negative, or neutral).
    """
    llm = ChatGoogleGenerativeAI(model=google_gemini_model) # Use a suitable Gemini model
    prompt = f"""
    You are an expert financial analyst. Your task is to analyze the sentiment of the following news article related to the stock {stock_symbol}.
    Determine if the overall sentiment is positive, negative, or neutral. 
    Provide the sentiment (positive, negative, or neutral) as well as recent news that let to specific sentiment in the output.
    Also state the reason for your sentiment analysis conclusion.
    Also provide long term outlook for the stock based on the news article and historical gain or loss since last 20 year or inception.

    News Article:
    {news_article}

    """

    messages = [HumanMessage(content=prompt)]
    response = llm.invoke(messages)
    return response.content
```

```python
# Capture the output in a variable and write onconsole in seperated lines
def get_stock_sentiment_output(ticker: str) -> str:
    """
    Analyzes the sentiment of a stock based on its news headlines and returns the output.

    Args:
        ticker: The stock ticker symbol (e.g., 'TSLA').

    Returns:
        A string containing the sentiment analysis output.
    """
    output = analyze_stock_sentiment(ticker, get_sorted_news_headlines(ticker))
    return output
```
