```python
# Define the prompt
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.schema import HumanMessage

def analyze_stock_sentiment(stock_symbol: str, news_article: str) -> str:
    """
    Analyzes the sentiment of a news article related to a specific stock.

    Args:
        stock_symbol: The ticker symbol of the stock (e.g., "CRCL").
        news_article: The text of the news article.

    Returns:
        A string indicating the sentiment (positive, negative, or neutral).
    """
    llm = ChatGoogleGenerativeAI(model="gemini-2.5-pro") # Use a suitable Gemini model
    prompt = f"""
    You are an expert financial analyst. Your task is to analyze the sentiment of the following news article related to the stock {stock_symbol}.
    Determine if the overall sentiment is positive, negative, or neutral. 
    Provide the sentiment (positive, negative, or neutral) as well as recent news that let to specific sentiment in the output.
    Also state the reason for your sentiment analysis conclusion.
    Also provide long term outlook for the stock based on the news article and historical gain or loss since last 20 year or inception.

    News Article:
    {news_article}

    """

    messages = [HumanMessage(content=prompt)]
    response = llm.invoke(messages)
    return response.content
```

```python
import requests

def get_finance_news(ticker, api_key):
    url = f"https://www.alphavantage.co/query"
    params = {
        "function": "NEWS_SENTIMENT",
        "tickers": ticker,
        "apikey": api_key
    }
    response = requests.get(url, params=params)
    if response.status_code == 200:
        data = response.json()
        return data.get("feed", [])
    else:
        return f"Error: {response.status_code}, {response.text}"

# Example usage
api_key = "your_alpha_vantage_api_key"
ticker = "AAPL"
news = get_finance_news(ticker, api_key)
for article in news:
    print(f"Title: {article['title']}")
    print(f"URL: {article['url']}")
    print()
```

```python
import requests
from bs4 import BeautifulSoup

# Specify the ticker symbol
ticker = "AAPL"

# Yahoo Finance news URL for the ticker
url = f"https://finance.yahoo.com/quote/{ticker}?p={ticker}&.tsrc=fin-srch"

# Fetch the webpage
response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')

# Extract news headlines
news_section = soup.find_all('h3', class_='Mb(5px)')
print(news_section)
for news in news_section:
    print(news)
    title = news.text
    link = "https://finance.yahoo.com" + news.a['href']
    print(f"Title: {title}")
    print(f"Link: {link}")
    print("-" * 50)
```
